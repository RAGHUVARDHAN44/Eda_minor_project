### Converting a column to numbers

### Grouping apps by category

`groupby()` collects app records with the same category for a comparison.

```python
category = apps.groupby("Category").agg(app_count=("App", "count"), mean_rating=("Rating", "mean"), median_installs=("Installs", "median")).sort_values("app_count", ascending=False)
```
